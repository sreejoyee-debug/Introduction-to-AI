```python
#problem2
from collections import deque

def water_jug_bfs():
    capacity = (4, 3)
    start = (0, 0)
    queue = deque([(start, [])])
    visited = {start}

    while queue:
        (a, b), path = queue.popleft()

        if a == 2 or b == 2:
            return path + [(a, b)]

        states = [
            (4, b),                         # Fill A
            (a, 3),                         # Fill B
            (0, b),                         # Empty A
            (a, 0),                         # Empty B
            (a - min(a, 3-b), b + min(a, 3-b)),  # A -> B
            (a + min(b, 4-a), b - min(b, 4-a))   # B -> A
        ]

        for state in states:
            if state not in visited:
                visited.add(state)
                queue.append((state, path + [(a, b)]))

    return None


solution = water_jug_bfs()

print("Minimum Steps:")
for i, state in enumerate(solution):
    print(f"Step {i}: Jug A = {state[0]}L, Jug B = {state[1]}L")
```
